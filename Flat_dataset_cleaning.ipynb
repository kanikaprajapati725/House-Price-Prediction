```python
def treat_price(price):
    if  type(price)== float:
      return price
    else:
      if 'Lac' in price:
       return round(float(price[0])/100,2)
      else:
        return round(float(price[0]),2)
```

```python
df['price']=df['price'].str.split(' ').apply(treat_price)
```
